```python
import csv
import re 

file= open("bank_statement_test.csv")

reader=csv.DictReader(file, delimiter=',')
```

```python
for row in reader:
    statement={"Acct_NO" : row['Account Number'], "Description": row["Transaction Description"], "Date" : row["Transaction Date"], "Type" : row["Transaction Type"], "Amount" : row["Transaction Amount"], "Balance" : row["Balance"] }
    print(statement)

def normalize(statement):
    if type(statement.get("Acct_NO")) != str:
        statement["Acct_NO"] = str(statement["Acct_NO"])
    if type(statement.get("Description")) != str:
        statement["Description"] = str(statement["Description"])
    if type(statement.get("Date")) != str:
        statement["Date"] = str(statement["Date"])
    if type(statement.get("Type")) != str:
        statement["Type"] = str(statement["Type"])
    try:
        if type(statement.get("Amount")) != float:
            statement["Amount"] = float(statement["Amount"])
    except ValueError:
        print(f"Error Cannot convert {statement.get("Amount")} to float")
    try:
        if type(statement.get("Balance")) != float:
            statement["Balance"] = float(statement["Balance"])
    except ValueError:
        print(f"Error Cannot convert {statement.get("Balance")} to float")
    
    
    
# def validate use regex for data valididdy

```
